Read Liquidator data

```python
# You code here
# 1. Read data from "LIQUIDATOR_202510" spreadsheet using Pandas
#    + Provide a schema when reading data file
liquidator_schema = {
    "REGISTER_NAME": str,
    "LIQ_NUM": str,
    "OFF_LIQ_NUM": str, 
    "LIQ_NAME": str,
    "LIQ_START_DT": str,
    "OFF_LIQ_START_DT": str,
    "LIQ_STATUS": str,
    "LIQ_SUSP_DT": str,
    "LIQ_ADD_LOCAL": str,
    "LIQ_ADD_STATE": str,
    "LIQ_ADD_PCODE": str,
    "LIQ_ADD_COUNTRY": str,
    "LIQ_FIRM": str
}

auditor_schema = {
    "REGISTER_NAME": str,
    "REG_AUD_NUM": str,
    "REG_AUD_NAME": str, 
    "REG_AUD_ACN": str,
    "REG_AUD_START_DT": str,
    "REG_AUD_STATUS": str,
    "REG_AUD_ADD_LOCAL": str,
    "REG_AUD_ADD_STATE": str,
    "REG_AUD_ADD_PCODE": str,
    "REG_AUD_ADD_COUNTRY": str,
}

# 2. Export to CSV as path 'destination/excel'
# We can use Pandas and enforce schema

df_auditor= pd.read_excel(excel_file_path, sheet_name = auditor_sheetname, header = 2)
df_liquidator= pd.read_excel(excel_file_path, sheet_name = liquidator_sheetname)
print(df_auditor.head())
print(df_liquidator.head())
```

Tranformation

```python
df_auditor.columns = df_auditor.columns.str.strip()

# Numeric columns
numeric_cols = ['REG_AUD_NUM', 'REG_AUD_ACN', 'REG_AUD_ADD_PCODE']
for col in numeric_cols:
    df_auditor[col] = pd.to_numeric(df_auditor[col], errors='coerce').fillna(0).astype(int)

# Data columns
date_cols = ['REG_AUD_START_DT', 'REG_AUD_SUSP_DT']
for col in date_cols:
    df_auditor[col] = pd.to_datetime(df_auditor[col], errors='coerce', dayfirst=True)

object_cols = df_auditor.select_dtypes(include='object').columns
df_auditor[object_cols] = df_auditor[object_cols].fillna('').astype(str)

print(df_auditor.head())
print("\nData types:")
print(df_auditor.dtypes)
```

```python

df_liquidator.columns = df_liquidator.columns.str.strip()

numeric_cols = ['LIQ_NUM', 'OFF_LIQ_NUM', 'LIQ_ADD_PCODE']
for col in numeric_cols:
    df_liquidator[col] = pd.to_numeric(df_liquidator[col], errors='coerce').fillna(0).astype(int)

# --- 3️⃣ Convert date columns ---
date_cols = ['LIQ_START_DT', 'OFF_LIQ_START_DT', 'LIQ_SUSP_DT']
for col in date_cols:
    df_liquidator[col] = pd.to_datetime(df_liquidator[col], errors='coerce', dayfirst=True)

# --- 4️⃣ Convert object columns → string ---
object_cols = df_liquidator.select_dtypes(include='object').columns
df_liquidator[object_cols] = df_liquidator[object_cols].fillna('').astype(str)

# --- 5️⃣ Check result ---
print(df_liquidator.head())
print("\nData types:")
print(df_liquidator.dtypes)
```

Export to CSV

```python
export_path = 'destination/excel'
os.makedirs(export_path, exist_ok=True)  

auditor_file = os.path.join(export_path, 'df_auditor_cleaned.csv')
df_auditor.to_csv(auditor_file, index=False)

liquidator_file = os.path.join(export_path, 'df_liquidator_cleaned.csv')
df_liquidator.to_csv(liquidator_file, index=False)

print(f"df_auditor exported to: {auditor_file}")
print(f"df_liquidator exported to: {liquidator_file}")
```
