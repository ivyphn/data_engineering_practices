```python
import pandas as pd
import json
```

1. Simple Structure

```python
simple_json_file_path = "simple_structure.json"

df_simple = pd.read_json(simple_json_file_path)
print(df_simple)
```

2. Multiple Level

```python
multiple_level_json_file_path = "multiple_level.json"

with open(multiple_level_json_file_path, "r") as f: 
    data = json.load(f)

df_multiple = pd.json_normalize(data["students"])
df_multiple["school_name"] = data["school_name"]
df_multiple["class"] = data["class"]
print(df_multiple)
```

3. Nested Levels

```python
nested_level_json_file_path = "nested_level.json"

with open(nested_level_json_file_path, "r") as f: 
    data = json.load(f)

df_nest = pd.json_normalize(data["students"])
df_nest["school_name"] = data["school_name"]
df_nest["class"] = data["class"]
print(df_nest)


```
