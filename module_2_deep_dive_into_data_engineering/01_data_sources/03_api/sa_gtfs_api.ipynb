```python
import requests
from zipfile import ZipFile
import os
from io import BytesIO
```

```python
# URLs
base_url = "http://gtfs.adelaidemetro.com.au/v1"
latest_version_number = "static/latest/version.txt"
latest_version_feed = "static/latest/google_transit.zip"

# binary data as response
request_headers={"Content-Type": "application/octet-stream"}

# Feed template
feed_url_template = "static/{version}/google_transit.zip"

# output
dest_path = 'destination'
os.makedirs(dest_path, exist_ok=True)
```

```python
# 1. Find the latest feed version
r = requests.get(f'{base_url}/{latest_version_number}')
print(r)

r.__dict__
```

```python
# 2. Download latest GTFS feed version
feed_url = f"{base_url}/{latest_version_feed}"
print(f"Latest feed URL: {feed_url}")

response = requests.get(url=feed_url, headers={"Content-Type": "binary/octet-stream"})
# Read content in chunks
with open("latest.zip", 'wb') as f:
    for chunk in response.iter_content(chunk_size=8192):
        f.write(chunk)

print("Latest version Download complete!")
```

```python
# # Dynamically download 10 versions from 930 to 939

for version in range(930, 940):
    version_feed_url = f"{base_url}/{feed_url_template.format(version=version)}"
    print(f"Downloading version {version}...")

    response = requests.get(version_feed_url)

    zip_path = f"{dest_path}/google_transit_{version}.zip"

    with open(zip_path, "wb") as f:
        for chunk in response.iter_content(chunk_size=8192):
            f.write(chunk)

    print(f"Version {version} is downloaded!")
```
